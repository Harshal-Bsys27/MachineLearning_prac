```python
#lets import libraries
import hashlib
from cryptography.hazmat.primitives.asymmetric import rsa,padding
from cryptography.hazmat.primitives import hashes
```

```python
#Step 2 : Create Transactions
transactions = [
    "A sends 10 BTC to B",
    "B sends 5 BTC to C",
    "C sends 2 BTC to D",
    "D sends 1 BTC to A"
]
```

```python
#STEP 3 : Generate Public and Private keys
private_key = rsa.generate_private_key(
    public_exponent=65537,
    key_size=2048
)
public_key = private_key.public_key()
```

```python
#Step 4 : Hash Transactions
def hash_data(data):
    return hashlib.sha256(data.encode()).hexdigest()
transaction_hashes = []

print("Transaction Hashes:\n")

for tx in transactions:
  tx_hash = hash_data(tx)
  transaction_hashes.append(tx_hash)
  print(f"{tx} --> {tx_hash}")
```

```python
# Step 5: Digital Signature (on first transaction)
message = transactions[0].encode()
signature = private_key.sign(
    message,
    padding.PSS(
        mgf=padding.MGF1(hashes.SHA256()),
        salt_length=padding.PSS.MAX_LENGTH
    ),
    hashes.SHA256()
)
print("\nDigital Signature Created for Transaction 1")
```

```python

# Step 6: Verify Signature
try:
    public_key.verify(
        signature,
        message,
        padding.PSS(
            mgf=padding.MGF1(hashes.SHA256()),
            salt_length=padding.PSS.MAX_LENGTH
        ),
        hashes.SHA256()
    )
    print("Signature Verified Successfully!")
except:
    print("Signature Verification Failed!")
```

```python
# Step 7: Block Hashing
# Simulating block data
previous_hash = "0000abc123"
block_data = "".join(transaction_hashes)
block_content = previous_hash + block_data
block_hash = hash_data(block_content)
print("\nBlock Hash:")
print(block_hash)
```

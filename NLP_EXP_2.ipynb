**EXPERIMENT NO 02 :  NLP LAB**

```python
import nltk
import string
import re
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
nltk.download('punkt')
nltk.download('stopwords')
#input text
text = "Natural Language processing is an pioneer for analysis in speech recoginition in 2025"
print("Original text")
print(text)
```

```python
nltk.download('punkt_tab')
tokens = word_tokenize(text)
print("\nTokens:")
print(tokens)
```

```python
from nltk.corpus import stopwords
# Filtration
stop_words = set(stopwords.words('english'))
filtered_words = []
```

```python
for word in tokens:
  word = word.lower()
  if word not in stop_words and word not in string.punctuation and word.isalpha():
    filtered_words.append(word)
print("\nFiltered Words:")
print(filtered_words)
```

```python
# Script Validation (English)
if re.fullmatch(r"[A-Za-z\s.,!?']+", text):
  print("\nScript Validation:")
  print("Valid English Script")
else:
  print("\nScript Validation:")
  print("Invalid or Mixed Script")
```
